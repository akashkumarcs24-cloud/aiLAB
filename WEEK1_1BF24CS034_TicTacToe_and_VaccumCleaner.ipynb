```python
states = [
    ("A", "DIRTY", "DIRTY"),
    ("A", "DIRTY", "CLEAN"),
    ("A", "CLEAN", "DIRTY"),
    ("A", "CLEAN", "CLEAN"),
    ("B", "DIRTY", "DIRTY"),
    ("B", "DIRTY", "CLEAN"),
    ("B", "CLEAN", "DIRTY"),
    ("B", "CLEAN", "CLEAN")
]

print("==========================================")
print(" 8 POSSIBLE VACUUM STATES")
print("==========================================")

for i, state in enumerate(states, 1):
    vacuum, room_a, room_b = state

    print("\nState", i)
    print("Vacuum Location :", vacuum)
    print("Room A :", room_a)
    print("Room B :", room_b)
    print("------------------------------------------")


def vacuum_cleaner(location, room_a, room_b):

    print("\n==========================================")
    print(" VACUUM CLEANER AGENT")
    print("==========================================")

    print("\nInitial State:")
    print("Vacuum Location :", location)
    print("Room A :", room_a)
    print("Room B :", room_b)

    step = 1

    while room_a == "DIRTY" or room_b == "DIRTY":

        print("\nStep", step)

        if location == "A":

            if room_a == "DIRTY":
                print("Current Room : A")
                print("Action : CLEAN")
                room_a = "CLEAN"

            else:
                print("Current Room : A")
                print("Action : MOVE RIGHT")
                location = "B"

        else:

            if room_b == "DIRTY":
                print("Current Room : B")
                print("Action : CLEAN")
                room_b = "CLEAN"

            else:
                print("Current Room : B")
                print("Action : MOVE LEFT")
                location = "A"

        print("New State:")
        print("Vacuum Location :", location)
        print("Room A :", room_a)
        print("Room B :", room_b)

        step += 1

    print("\n==========================================")
    print(" GOAL REACHED")
    print("==========================================")

    print("Both rooms are CLEAN!")

    print("\nFinal State:")
    print("Vacuum Location :", location)
    print("Room A :", room_a)
    print("Room B :", room_b)


print("\n\n==========================================")
print(" SELECT INITIAL STATE")
print("==========================================")

choice = int(input("Enter state number (1-8): "))

if 1 <= choice <= 8:
    location, room_a, room_b = states[choice - 1]
    vacuum_cleaner(location, room_a, room_b)
else:
    print("Invalid state number!")
```

```python
# Tic-Tac-Toe using Game Tree / Minimax

WIN_LINES = [
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),
    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),
    (0, 4, 8),
    (2, 4, 6)
]


# Display board
def print_board(board):
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


# Check winner
def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != " " and board[a] == board[b] == board[c]:
            return board[a]

    return None


# Check whether board is full
def is_full(board):
    return " " not in board


# Minimax algorithm
def minimax(board, turn):

    win = winner(board)

    # Computer wins
    if win == "O":
        return 1

    # Human wins
    if win == "X":
        return -1

    # Draw
    if is_full(board):
        return 0

    # Computer's turn
    if turn == "O":

        best_score = -100

        for i in range(9):

            if board[i] == " ":

                board[i] = "O"

                score = minimax(board, "X")

                board[i] = " "

                best_score = max(best_score, score)

        return best_score

    # Human's turn
    else:

        best_score = 100

        for i in range(9):

            if board[i] == " ":

                board[i] = "X"

                score = minimax(board, "O")

                board[i] = " "

                best_score = min(best_score, score)

        return best_score


# Find best move for computer
def computer_move(board):

    best_score = -100
    best_move = None

    for i in range(9):

        if board[i] == " ":

            board[i] = "O"

            score = minimax(board, "X")

            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    return best_move


# Main game
def play_game():

    board = [" "] * 9

    print("TIC-TAC-TOE")
    print("You = X")
    print("Computer = O")

    while True:

        print_board(board)

        # Human move
        move = int(input("Enter position (1-9): ")) - 1

        if move < 0 or move > 8:
            print("Invalid position!")
            continue

        if board[move] != " ":
            print("Position already occupied!")
            continue

        board[move] = "X"

        # Check human win
        if winner(board) == "X":
            print_board(board)
            print("You Win!")
            break

        if is_full(board):
            print_board(board)
            print("Draw!")
            break

        # Computer move
        move = computer_move(board)
        board[move] = "O"

        print("Computer chose:", move + 1)

        # Check computer win
        if winner(board) == "O":
            print_board(board)
            print("Computer Wins!")
            break

        if is_full(board):
            print_board(board)
            print("Draw!")
            break


play_game()
```
